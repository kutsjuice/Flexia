# PositionLinearActuator2D

```julia
@variables  _xi _yi _θi _xj _yj _θj

@variables xci yci xcj ycj  
@variables xdi ydi xdj ydj  
        

# Local axis angles (intrinsic constants)
αi = atan(ydi, xdi)
αj = atan(ydj, xdj)

# Rotation matrices
Ri = [cos(_θi) -sin(_θi); sin(_θi) cos(_θi)]
Rj = [cos(_θj) -sin(_θj); sin(_θj) cos(_θj)]

xpi = _xi + xci * cos(_θi) - yci * sin(_θi)
ypi = _yi + xci * sin(_θi) + yci * cos(_θi)

xpj = _xj + xcj * cos(_θj) - ycj * sin(_θj)
ypj = _yj + xcj * sin(_θj) + ycj * cos(_θj)

# direction in terms of first body in global CS
N_xni = xdi*cos(_θi) - ydi*sin(_θi)
N_yni = xdi*sin(_θi) + ydi*cos(_θi)

distance = (xpj-xpi) * N_xni + (ypj-ypi) * N_yni
```

```julia
q = [_xi, _yi, _θi, _xj, _yj, _θj]
jac = Symbolics.gradient(distance, q);
simplify(jac[6])
```

```julia
expr1 = (ypj - _yi)*(N_xni) - (xpj - _xi) * N_yni
```
